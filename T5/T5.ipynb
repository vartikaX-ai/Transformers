```python
from transformers import T5Tokenizer,TFT5ForConditionalGeneration
```

```python
tokenizer = T5Tokenizer.from_pretrained("google/flan-t5-base")
model = TFT5ForConditionalGeneration.from_pretrained("google/flan-t5-base")
```

Summarization

```python
text_summary = """Write a detailed summary of the following text::

Artificial intelligence is transforming many industries around the world. Machine learning enables computers to learn patterns from data, while deep learning uses neural networks to solve complex problems. Generative AI has expanded these capabilities by allowing models to generate text, images, audio, and code. These technologies are increasingly being used in healthcare, finance, education, transportation, and entertainment."""
```

```python
inputs_summary = tokenizer(text_summary,return_tensors="tf",truncation=True)
```

```python
output_summary = model.generate(input_ids = inputs_summary["input_ids"],attention_mask=inputs_summary["attention_mask"],num_beams=4,max_length=150,min_length=40)
result_summary = tokenizer.decode(output_summary[0],skip_special_tokens=True)
print(result_summary)
```
