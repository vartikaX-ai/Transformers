```python
import numpy as np
import tensorflow as tf
from transformers import TFAutoModelForCausalLM,AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
model = TFAutoModelForCausalLM.from_pretrained("openai-community/gpt2")
```

```python
if(tokenizer.pad_token is None):
  tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.pad_token_id
```

```python
prompt =  """
Artificial intelligence is transforming modern healthcare.
It helps doctors by
"""
inputs = tokenizer(prompt,return_tensors = "tf")
print(inputs["input_ids"])
print(inputs["attention_mask"])
```

```python
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0].numpy().tolist())
print(tokens)
```

```python
#Greedy Decoding
greedy_output = model.generate(
    input_ids = inputs["input_ids"],
    attention_mask = inputs["attention_mask"],
    max_new_tokens = 50,
    do_sample = False,
    pad_token_id = tokenizer.pad_token_id
)

greedy_text = tokenizer.decode(greedy_output[0],skip_special_tokens=True)
print(greedy_text)
```

```python
#Temperature
temperature_output = model.generate(
    input_ids = inputs["input_ids"],
    attention_mask = inputs["attention_mask"],
    max_new_tokens=50,
    do_sample = True,
    temperature=0.7,
    pad_token_id = tokenizer.pad_token_id
)

temperature_text = tokenizer.decode(temperature_output[0],skip_special_tokens=True)
print(temperature_text)
```

```python
#Top-k sampling
topk_output = model.generate(
    input_ids = inputs["input_ids"],
    attention_mask = inputs["attention_mask"],
    max_new_tokens=50,
    do_sample = True,
    temperature=0.7,
    top_k=50,
    pad_token_id = tokenizer.pad_token_id
)

topk_text = tokenizer.decode(topk_output[0],skip_special_tokens=True)
print(topk_text)
```

```python
topp_output = model.generate(
    input_ids = inputs["input_ids"],
    attention_mask = inputs["attention_mask"],
    max_new_tokens = 50,
    do_sample = True,
    temperature = 0.7,
    top_p = 0.85,
    pad_token_id = tokenizer.pad_token_id
)

topp_text = tokenizer.decode(topp_output[0],skip_special_tokens=True)
print(topp_text)
```
